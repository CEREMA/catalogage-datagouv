```python
import requests
import pandas as pd
import time

DATAGOUV_API = "https://www.data.gouv.fr/api/1"

def fetch_datasets_df(page_size=1000, max_pages=None, sleep=0.3) -> pd.DataFrame:
    records = []
    page = 1
    session = requests.Session()

    while True:
        resp = session.get(
            f"{DATAGOUV_API}/datasets/",
            params={"page_size": page_size, "page": page},
            timeout=60,
        )
        resp.raise_for_status()
        data = resp.json()
        items = data.get("data", [])
        if not items:
            break

        for ds in items:
            records.append({
                "id": ds["id"],
                "title": ds.get("title", ""),
                "description": (ds.get("description") or "")[:1000],
                "org": (ds.get("organization") or {}).get("name", ""),
                "org_id": (ds.get("organization") or {}).get("id", ""),
                "tags": ds.get("tags") or [],
                "views": (ds.get("metrics") or {}).get("views", 0),
                "ressources_nb": len(ds.get("resources") or []),
                "ressources_texte": ";".join([f"{i+1}. {d['title']} - {d['description']}" for i, d in enumerate(ds.get("resources") or [])]),
                "downloads": (ds.get("metrics") or {}).get("resources_downloads", 0),
                "license": ds.get("license"),
                "frequency": ds.get("frequency"),
                "last_modified": ds.get("last_modified"),
            })

        print(f"Page {page}: +{len(items)} ({len(records)} total)")
        if not data.get("next_page") or (max_pages and page >= max_pages):
            break
        page += 1
        time.sleep(sleep)

    return pd.DataFrame(records)
```

```python
!pip install sentence-transformers pyyaml

import yaml
import numpy as np
from sentence_transformers import SentenceTransformer, util
```

```python
def load_categories(path="categories.yaml"):
    with open(path, "r", encoding="utf-8") as f:
        cats = yaml.safe_load(f)

    flat = []  # liste plate (catégorie, sous-catégorie, texte de référence)
    for cat in cats:
        for sub in cat.get("subcategories", []):
            flat.append({
                "category": cat["slug"],
                "subcategory": sub["slug"],
                "label": f"{cat['label']} > {sub['label']}",
                "text": f"{cat['description']}. {sub['description']}",
            })
    return flat
```

```python
def classify_df(df, categories_path="categories.yaml",
                 model_name="paraphrase-multilingual-mpnet-base-v2",
                 threshold=0.25):
    model = SentenceTransformer(model_name)
    cats = load_categories(categories_path)

    # Embeddings des catégories (une seule fois)
    cat_texts = [c["text"] for c in cats]
    cat_embeddings = model.encode(cat_texts, convert_to_tensor=True)

    # Embeddings des datasets (titre + description + tags)
    doc_texts = (
        df["title"].fillna("") + ". " +
        df["description"].fillna("") + ". " +
        df["tags"].apply(lambda t: " ".join(t) if isinstance(t, list) else "")
    ).tolist()
    doc_embeddings = model.encode(doc_texts, convert_to_tensor=True,
                                   batch_size=64, show_progress_bar=True)

    # Similarité cosinus dataset x catégorie
    sims = util.cos_sim(doc_embeddings, cat_embeddings).cpu().numpy()

    best_idx = sims.argmax(axis=1)
    best_score = sims.max(axis=1)

    df = df.copy()
    df["category"] = [cats[i]["category"] if s >= threshold else "" for i, s in zip(best_idx, best_score)]
    df["subcategory"] = [cats[i]["subcategory"] if s >= threshold else "" for i, s in zip(best_idx, best_score)]
    df["category_score"] = best_score
    return df
```

```python

dsdatagouv = pd.read_csv("export-dataset-20260817-055836.csv", sep=";")
```

```python
# le catalogue des datasets contient aussi ceux qui sont archivsé : on ne va pas les traiter (on passe ainsi de 130k à 73k datasets)
dsdatagouv=dsdatagouv[dsdatagouv["archived"] == "False"]
```

```python
keywords = [
    "transport", "mobilité", "route", "passager", "fret", "parking", "stationnement", "voirie", "accessibilité", "enquête", "horaire", "calendrier", 
    "vélo", "trafic", "piéton", "voiture", "véhicule", "carburant", "accident", "sécurité routière", "train", "transport public", "avion", "fluvial", "maritime",
    "covoiturage", "autopartage", "bruit", "qualité de l'air"
]
```

```python
# Construire un pattern regex : mot1|mot2|mot3...
pattern = "|".join(re.escape(kw) for kw in keywords)

# Concaténer title + description pour chercher dans les deux
text = (
    dsdatagouv["title"].fillna("") + " " + dsdatagouv["description"].fillna("")
)
```

```python
mask = text.str.contains(pattern, case=False, na=False, regex=True)
dsdatagouv_mob = dsdatagouv[mask].copy()
```
